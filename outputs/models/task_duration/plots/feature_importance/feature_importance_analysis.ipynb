```python
from PIL import Image
from pathlib import Path

INPUT_DIR = Path(".").resolve()
OUTPUT_DIR = INPUT_DIR / "combined"
OUTPUT_DIR.mkdir(exist_ok=True)

# -------- Collect images --------
images = list(INPUT_DIR.glob("*.png"))

# Index files by dataset name
groups = {}

for img_path in images:
    name = img_path.name
    if "log_duration" not in name:
        continue

    dataset = name.split("_log_duration")[0]
    groups.setdefault(dataset, {})["copy" if "copy" in name else "main"] = img_path

# -------- Combine images --------
for dataset, files in groups.items():
    if "copy" not in files or "main" not in files:
        print(f"Skipping {dataset}: missing pair")
        continue

    left_img = Image.open(files["copy"])
    right_img = Image.open(files["main"])

    # Ensure same height
    h = min(left_img.height, right_img.height)
    left_img = left_img.resize(
        (int(left_img.width * h / left_img.height), h)
    )
    right_img = right_img.resize(
        (int(right_img.width * h / right_img.height), h)
    )

    # Create combined canvas
    combined = Image.new(
        "RGB",
        (left_img.width + right_img.width, h),
        color=(255, 255, 255),
    )

    combined.paste(left_img, (0, 0))
    combined.paste(right_img, (left_img.width, 0))

    # Save
    out_path = OUTPUT_DIR / f"{dataset}_feature_importance_combined.png"
    combined.save(out_path)
    print(f"Saved: {out_path}")
```
