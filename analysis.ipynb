```python
import pandas as pd
```

```python
df = pd.read_csv("outputs\preprocessed\merged.csv")
```

```python
df
```

```python
df['total_duration_sec'].mean() /3600
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

# Plot relative_start
sns.histplot(
    df['relative_start'],
    bins=40,
    color='steelblue',
    label='Relative Start',
    kde=True,
    alpha=0.6,
)

plt.xlim(0, 1.2)
plt.xlabel("Relative Start Sequence")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
import numpy as np
df['relative_complete_log'] = np.log1p(df['relative_complete'])
```

```python
plt.figure(figsize=(8, 5))

# Plot relative_start
sns.histplot(
    df['total_duration_sec']/3600,
    bins=1000,
    color='steelblue',
    label='Relative Start',
    kde=True,
    alpha=0.6,
)

plt.xlim(0, 500)
plt.xlabel("Relative Start Sequence")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
